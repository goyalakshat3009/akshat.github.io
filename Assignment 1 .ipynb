```python
# Q.5
a = [1, 4, 2, 2, 3]
n = len(a)

if n == 0:
    print("Empty list")
else:
    # Mean
    total = 0
    for x in a:
        total += x
    mean = total / n

    # Sorting
    for i in range(n):
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                temp = a[j]
                a[j] = a[j + 1]
                a[j + 1] = temp

    # Median
    if n % 2 == 0:
        median = (a[n/2-1] + a[n/2]) / 2
    else:
        median = a[n//2]

    # Mode
    freq = {}
    for x in a:
        if x in freq:
            freq[x]+=1
        else:
            freq[x]=1

    max_count = 0
    mode = []

    for key in freq:
        if freq[key]>max_count:
            max_count = freq[key]
            mode = [key]
        elif freq[key] == max_count:
            mode.append(key)

    #Output
    print("Sorted List =", a)
    print("Mean =", mean)
    print("Median =", median)

    if max_count == 1:
        print("Mode = No mode")
    else:
        print("Mode =", mode)
```

```python
# Q.6
a=[10,20,30]
total=0

for x in a:
  total+=x
mean = total/len(a)

var=0
for x in a:
  var += (x-mean)**2
sd = (var/len(a))**0.5
print("Standard deviation is :", sd)
```

```python
# Q.8
y = [3,5,2]
x = [2,5,4]

mae=0
mse=0

for i in range(len(y)):
    mae += abs(y[i]-x[i])
    mse += (y[i]-x[i])**2

print("MAE =", mae/len(y))
print("MSE =", mse/len(y))
```
